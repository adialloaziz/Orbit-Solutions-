```python
y_fp = (1/(2*np.pi))*np.ones_like(z_centers)+ 1e-11*np.sin(z_centers)
Y = []
solutions = [] #For stability analysis
Ivals = np.linspace(0.1*Ic, Ic, 9, endpoint=False)
for k, I in enumerate(Ivals):
    model.I = I
    y_fp = fixed_point(model.stationary_eq, y_fp,method='del2', xtol=1e-7, maxiter=10000)
    Y.append(y_fp)
    sol = solve_ivp(fun=model.dydt, t_span=(0, 60*T), y0=y_fp,
                     method='BDF', rtol = 1e-8, atol=1e-10, jac=J, t_eval=np.linspace(0, 60*T, 1000))
    solutions.append(sol)
    print(k)

plt.plot(z_centers, Y[0], color='red', linewidth=1.5, label=r'Fixed Point Solution')
plt.xlabel('z',fontsize =18)
plt.ylabel(r'$\rho(z)$', fontsize = 18)
plt.title('Fixed Point Solution of the Density')
plt.legend()
plt.grid()
plt.show()
```

```python
#Ploting the branch of solutions
plt.figure(figsize=(8,6))
first_moment = [np.trapezoid(z_centers*sol.y[:,-1], z_centers) for sol in solutions]
norm_inf_y = [np.linalg.norm(sol.y[:,-1], 2) for sol in solutions]
plt.plot(Ivals, norm_inf_y, marker='o')
plt.xlabel('I', fontsize=18)
plt.ylabel(r'$\|y\|_{\infty}$', fontsize=18)
plt.title('Branch of Fixed Point Solutions')
plt.legend()
plt.grid()
plt.show()
```
